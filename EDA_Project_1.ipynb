```python
# Cleaning

# EDA-4: Identify NaN values
df = df.replace(" ?", None )
print(df.isnull().sum())
print("_________________________")
# Fill with median
df["Age"] = df["Age"].fillna(df["Age"].median())
print(df)
```

```python
# Cleaning

# EDA-4: Identify NaN values
df = df.replace(" ?", None )
print(df.isnull().sum())
print("_________________________")
# Fill with mode
df["Occupation"] = df["Occupation"].fillna(df["Occupation"].mode())
print(df)
```

```python
#import pandas as pd
#df = pd.read_csv("adult.csv")

# Cleaning
# Approach-1: Delete the records (rows)
df = df.dropna()
print(df)
print("_________________________")

# Approach-2: Delete the columns
df = df.dropna(axis=1)
print(df)
print("_________________________")
```

```python
df['Age']=df['Age'].bfill()
print(df)
print("_________________________")
df['Workclass'] = df['Workclass'].bfill()
print(df)
df['Ocuupation'] = df['Occupation'].bfill()
print(df)
print("_________________________")
df['Native Country'] = df['Native Country'].bfill()
print(df)
```

```python
df['Age']=df['Age'].ffill()
print(df)
print("_________________________")
df['Workclass'] = df['Workclass'].ffill()
print(df)
df['Ocuupation'] = df['Occupation'].ffill()
print(df)
print("_________________________")
df['Native Country'] = df['Native Country'].ffill()
print(df)
```

```python
df['Age']=df['Age'].interpolate()
print(df)
print("_________________________")
df['Workclass'] = df['Workclass'].interpolate()
print(df)
df['Ocuupation'] = df['Occupation'].interpolate()
print(df)
print("_________________________")
df['Native Country'] = df['Native Country'].interpolate()
print(df)
```

```python
#Detecting Outliers
Q1 = df['Age'].quantile(0.25)
Q3 = df['Age'].quantile(0.75)

IQR = Q3-Q1

print('Q1',Q1)
print('Q3',Q3)
print('IQR',IQR)

lower_bound = (Q1-1.5*IQR)
upper_bound = (Q3+1.5*IQR)
print(lower_bound)
print(upper_bound)
```

```python
#Identify IQR Outliers

iqr_outliers = df[(df['Age'] < lower_bound) | (df['Age'] > upper_bound)]
print("IQR Outliers:",len(iqr_outliers))
```

```python
#Z-Score Method
from scipy.stats import zscore
from scipy.stats.mstats import winsorize
df['Age_zscore'] = zscore(df['Age'])
z_outliers = df[df['Age_zscore'].abs() > 3]
print("Z-Score Outliers:", len(z_outliers))
```

```python
#Z-Score Trimming
df_trimmed = df[df['Age_zscore'].abs() <=3]
print(df_trimmed)
```

```python
#Winsorization
df['Age_winsorized'] = winsorize(df['Age'], limits=[0.05,0.5])
print(df.head())
```

```python
df['EducationNum_zscore'] = zscore(df['EducationNum'])
z_outliers = df[df['EducationNum_zscore'].abs() > 3]
print("Z-Score Outliers:", len(z_outliers))
```

```python
df_trimmed = df[df['EducationNum_zscore'].abs() <=3]
print(df_trimmed)
```

```python
#Winsorization
df['EducationNum_winsorized'] = winsorize(df['EducationNum'], limits=[0.05,0.5])
print(df.head())
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(10, 5))

sns.countplot(
    data=df,
    x="Education",
    hue="Income",
    order=[" HS-grad", " Some-college", " Bachelors", " Masters", " Doctorate"],
    palette=["#4c72b0", "#55a868"],
)

plt.title("Chart 1 (Bar Chart): Income Count by Education Level", fontsize=13)
plt.xlabel("Education Level", fontsize=11)
plt.ylabel("Number of People", fontsize=11)
plt.legend(title="Income Level")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(6, 6))

income_counts = df["Income"].value_counts()

# Create a donut chart using pie() with a center cutout
plt.pie(
    income_counts,
    labels=income_counts.index,
    autopct="%1.1f%%",
    startangle=90,
    colors=["#4c72b0", "#55a868"],
    wedgeprops=dict(width=0.4, edgecolor="white", linewidth=2),
)

plt.title("Chart 2 (Donut Chart): Proportion of <=50K vs >50K Earners", fontsize=13)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df.sample(n=1000, random_state=42),  # Sample 1000 points for clean viewing
    x="Age",
    y="Hours per Week",
    hue="Income",
    palette=["#4c72b0", "#e74c3c"],
    alpha=0.6,
    s=50,
)

plt.title("Chart 3 (Scatter Plot): Individual Age vs. Hours Worked per Week", fontsize=13)
plt.xlabel("Age (Years)", fontsize=11)
plt.ylabel("Hours Worked per Week", fontsize=11)
plt.legend(title="Income Bracket")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 5))

pivot_hours = df.pivot_table(
    index="Education",
    columns="Marital Status",
    values="Hours per Week",
    aggfunc="mean",
)

sns.heatmap(
    pivot_hours,
    cmap="YlGnBu",
    cbar_kws={"label": "Hours / Week"},
)

plt.title("Chart 4 (Heatmap): Average Working Hours per Week", fontsize=13)
plt.xlabel("Marital Status", fontsize=11)
plt.ylabel("Education Level", fontsize=11)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 5))

# Calculate average weekly hours worked across each age year for both income brackets
age_trend = df.groupby(["Age", "Income"])["Hours per Week"].mean().reset_index()

sns.lineplot(
    data=age_trend,
    x="Age",
    y="Hours per Week",
    hue="Income",
    palette=["#4c72b0", "#2ecc71"],
    linewidth=2.5,
)

plt.title("Chart 5 (Line Chart): Average Working Hours Trend Across Age", fontsize=13)
plt.xlabel("Age (Years)", fontsize=11)
plt.ylabel("Average Hours Worked per Week", fontsize=11)
plt.legend(title="Income Bracket")
plt.tight_layout()
plt.show()
```
